Preprocess the data

```python
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
label_encoder = LabelEncoder()
data['label'] = label_encoder.fit_transform(data['label'])
```

```python
# Separate features and labels
X = data.drop('label', axis=1)
y = data['label']
print(x)
print(y)
```

```python
from sklearn.impute import SimpleImputer
# Impute missing values
imputer = SimpleImputer(strategy='mean')
X_imputed = imputer.fit_transform(X)
```

```python
# Scale using MinMaxScaler (required for chi2)
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X_imputed)
```

Feature Selection

```python
from sklearn.feature_selection import SelectKBest, chi2
# Step 6: Feature selection using Chi2
selector = SelectKBest(score_func=chi2, k=20)
X_selected = selector.fit_transform(X_scaled, y)

feature_names = X.columns
scores = selector.scores_
selected_indices = selector.get_support(indices=True)
selected_features = feature_names[selected_indices]
selected_scores = scores[selected_indices]

# Create DataFrame for plotting
feature_scores = pd.DataFrame({
    'Feature': selected_features,
    'Chi2 Score': selected_scores
}).sort_values(by='Chi2 Score', ascending=True)

plt.figure(figsize=(12, 6))
plt.barh(feature_scores['Feature'], feature_scores['Chi2 Score'], color='skyblue')
plt.xlabel("Chi2 Score")
plt.title("Top 20 Features Selected by Chi-Square Test")
plt.tight_layout()
plt.show()
```

```python
selector = SelectKBest(score_func=chi2, k=20)
X_selected = selector.fit_transform(X_scaled, y)
selected_features = selector.get_feature_names_out(X.columns)
print(X_selected)
```

Model Selection

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_selected, y, test_size=0.2, random_state=42)
```

Performance Matrix of different algorithms

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import numpy as np
models={
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Support Vector Machine": SVC(kernel='rbf', random_state=42),
    "K-Nearest Neighbors": KNeighborsClassifier(n_neighbors=5),
    "RandomForestClassifier":RandomForestClassifier(n_estimators=100,random_state=42)
}
for name, model in models.items():
    print(f"\nResults for: {name}")
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    print("Accuracy:", accuracy_score(y_test, y_pred))
    print("Classification Report:\n", classification_report(
        y_test, y_pred, target_names=np.unique(y).astype(str)))
    print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

Decision Tree Visualization

```python
# Train an unpruned Decision Tree
full_tree = DecisionTreeClassifier(random_state=42)
full_tree.fit(X_train, y_train)

# Visualize the full tree
plt.figure(figsize=(24, 12))
plot_tree(full_tree,
          filled=True,
          feature_names=selector.get_feature_names_out(X.columns),
          class_names=[str(cls) for cls in np.unique(y)],
          rounded=True,
          fontsize=10)
plt.title("Unpruned Decision Tree")
plt.show()
```

Pruned Decision Tree

```python
# Train a pruned decision tree
pruned_tree = DecisionTreeClassifier(max_depth=3, random_state=42)
pruned_tree.fit(X_train, y_train)

# Plot the tree
plt.figure(figsize=(20, 10))
plot_tree(pruned_tree,
          filled=True,
          feature_names=selector.get_feature_names_out(X.columns),
          class_names=[str(cls) for cls in np.unique(y)],
          rounded=True,
          fontsize=10)
plt.title("Pruned Decision Tree (max_depth=4)")
plt.show()
```
